
My approach for this problem is instead of maintaining a seperate datastructure , what we can instead
do is read the input file and in parllell update the output file, first read one line store it in a local data structure
then write this in the output file , and continue .... return when the writing is finished, this way we only ever use a data structure
of size that of the fix-width format line length (ex 60)

```python
import os
def convert_fasta(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        sequence = ""
        for line in infile:
            line = line.strip()  
            if line.startswith(">"):
                
                if sequence:
                    outfile.write(sequence + "\n")
                # header line
                outfile.write(line + "\n")
                
                sequence = ""
            else:
                sequence += line
        if sequence:
            outfile.write(sequence + "\n")
            
input_path = os.path.join("BEE_Data", "q3_data", "multiline_input.fasta")
output_path = os.path.join("BEE_Data", "q3_data", "output.fasta")

convert_fasta(input_path, output_path)

```

The output is written to the file inside q3_data
