Looking at the data, it looks like the protien binded region is already mentioned,
we just have to verify it, by recentering the fragment centres wrt to centre of th  binded region we
should expect the reverse V plot to appear at the origin (X=0) where the protien is binded to the DNA (where neclease cant cleave)

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
import pandas as pd

path = os.path.join("BEE_Data", "q5_data", "mapped.bed")

heatmap_counts = defaultdict(int)

with open(path) as f:
    for line in f:
        cols = line.strip().split('\t')
        if len(cols) < 11:
            continue

        
        s1 = int(cols[2])
        e1 = int(cols[3])
        C1 = (s1 + e1) / 2

        s2 = int(cols[8])
        e2 = int(cols[9])
        C2 = (s2 + e2) / 2

        X = C2 - C1
        Y = e2 - s2 #i think column 12 already has this data, but i am computing manually anyways

        heatmap_counts[(X, Y)] += 1
```

```python
data = defaultdict(lambda: defaultdict(int))

for (X, Y), count in heatmap_counts.items():
    data[X][Y] = count


df = pd.DataFrame(data).fillna(0)  


plt.figure(figsize=(10, 8))
sns.heatmap(df.astype(int), cmap="YlGnBu", annot=True, fmt="d", linewidths=0.5)

plt.title("Heatmap of Protein Binding Fragments")
plt.xlabel("Fragment Length (Y)")
plt.ylabel("Difference Between Centers (X)")
plt.show()
```
